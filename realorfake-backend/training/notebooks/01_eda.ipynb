# 01 — Exploratory Data Analysis
CIFAKE dataset statistics, class balance, image size distribution, sample visualizations.

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from PIL import Image

DATA_ROOT = Path('data/cifake')  # adjust as needed
```

```python
# Count images per split and class
for split in ['train', 'val', 'test']:
    for cls in ['real', 'ai_generated']:
        p = DATA_ROOT / split / cls
        if p.exists():
            n = len(list(p.rglob('*.jpg')))
            print(f'{split}/{cls}: {n} images')
```
